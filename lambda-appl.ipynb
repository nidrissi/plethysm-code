### Sanity check

Check that $s_\lambda = \det(e_{\lambda'_i - i + j})_{1 \leq i, j \leq \ell(\lambda')}$, as expected.

```python
# chk[\[Lambda]_List, n_Integer] := ...; chk[{2, 1}, 4], chk[{3, 2, 1}, 4], chk[{4, 2, 1}, 5]
def chk(la, n):
    lt = Partition(la).conjugate()
    E = lambda k: e[k] if k >= 0 else e.zero()
    jt = matrix(e, len(lt), len(lt), lambda i, j: E(lt[i] - i + j)).det()
    return jt.expand(n) == s[la].expand(n)

chk([2, 1], 4), chk([3, 2, 1], 4), chk([4, 2, 1], 5)
```

## Orthogonality

Character of the regular representation, in two variables.

```python
# Table[reg2[n, x, y], {n, 4}] // Column
def reg(n):
    return sum(tensor([s[la], s[la]]) for la in Partitions(n))

def reg2(n):
    return sum(tensor([p[la], p[la]]) / la.centralizer_size() for la in Partitions(n))

for n in range(1, 5):
    print(reg2(n))
```

```python
# Table[conv[p][reg[n, x, y]] == reg2[n, x, y], {n, 5}] // Reduce
all(P2(reg(n)) == reg2(n) for n in range(1, 6))
```

```python
# Table[conv[s][reg2[n, x, y]] == reg[n, x, y], {n, 5}] // Reduce
all(S2(reg2(n)) == reg(n) for n in range(1, 6))
```

### Stable twisted cohomology of automorphism groups of free groups

We have: $\mathcal{Q}(q, p) = \mathrm{triv}_p$ if $q \in \{0, 1\}$ and $p \geq 1$, and $0$ otherwise.
Moreover $\mathcal{H} = \omega(\mathcal{P}) = \omega(\mathcal{F}(S(\mathcal{Q})))$ is the involution applied to the
saturation of $\mathcal{Q}$. Unlike the Albanese cohomology, this is infinite dimensional in any fixed degree
(because $\mathcal{Q}(1, 0)$ contains an element of degree $0$), so we can only compute the truncation up to some
arity.

```python
# TableForm[Table[{chQ[d, x, y]}, {d, 0, 6}], ...]
EH = tensor([e, h])
table([("d", "ch(Q) in degree d")] + [(d, EH(chQ(d))) for d in range(7)])
```

```python
# chH_upto[2, 5, x, y] // printPoly[S, x, y]
show_poly(chH_upto_input_arity(2, 5))
```

The characters of $\mathcal{P}$ and $\mathcal{H}$ in degree $d$, output arity $m$, input arity $n$ (nonzero entries
only). In the Mathematica version, the term $S_{\varnothing,1^2}^{\oplus 2}$ also appeared for $n = 2$ and every
$m \geq 1$, because the truncation in `chP[d, m, n]` did not remove the terms of output arity $0$.

```python
# TableForm[Map[printPoly[S, x, y], Table[chH[2, m, n, x, y], {m, 0, 5}, {n, 0, 5}], {2}], ...]
display(Markdown("\n".join(f"- $m = {m}$, $n = {n}$: ${character_latex(chH(2, m, n))}$"
                           for m in range(6) for n in range(6) if chH(2, m, n))))
```

### Albanese cohomology for $IA_\infty$

We have: $\mathcal{Q}'(q, p) = \mathrm{triv}_p[p]$ if $q = 0$ and $p \geq 1$, $\mathrm{triv}_p[p - 1]$ if $q = 1$ and
$p \geq 2$, and $0$ otherwise. Moreover $\mathcal{H}' = \omega(\mathcal{P}') = \omega(S(\mathcal{Q}'))$ is the
involution applied to the saturation of $\mathcal{Q}'$.

The following function gives the character of $\mathcal{Q}'$ in homological degree $d$:

```python
# TableForm[Table[{chQ_prime[d, x, y]}, {d, 0, 6}], ...]
H2 = tensor([h, h])
table([("d", "ch(Q') in degree d")] + [(d, H2(chQ_prime(d))) for d in range(7)])
```

Now, the saturation $\mathcal{P}' = S(\mathcal{Q}')$ has character
$\sum_{n \geq 1} h_n(x) \circ (\mathrm{ch}(\mathcal{Q}'), 0)$. The following function gives its part in degree
exactly $d$ (without the $(-\hbar)^d$).

```python
# TableForm[Table[{d, printPoly[S, x, y][chP_prime[d, x, y]]}, {d, 3}], ...]
display(Markdown("\n".join(f"- $d = {d}$: ${character_latex(chP_prime(d))}$" for d in range(1, 4))))
```

### Lazy series

The characters are also lazy power series in a variable $t$ recording the degree $d$: their coefficients are computed when needed, then cached, and `chP`, `chH`, `chP_prime`, ... read their values off these series. Applying a function to each coefficient gives other lazy series, for example the dimensions of $\mathcal{H}'_d$ computed above from the table. Displaying a series computes its first seven coefficients, here $d \leq 7$.

```python
L.<t> = LazyPowerSeriesRing(QQ)
L(lambda d: sum(c * la.dimension() * mu.dimension() for (la, mu), c in chH_prime_series[d]), valuation=1)
```

For $\mathcal{H}$, each degree has infinitely many terms, so the coefficient of $t^d$ is a lazy symmetric function graded by total arity $m + n$: in degree $d$, its component of total arity $m + n$ is the character in output arity $m$ and input arity $n$. No bound on the arity has to be chosen in advance:

```python
d = 1
display(Markdown("\n".join(f"- $m = {m}$, $n = {m + d}$: ${character_latex(chH_series[d][2 * m + d])}$"
                           for m in range(5))))
```

Since $\mathcal{Q}$ is $\mathcal{Q}'$ plus $h_1(x) h_1(y)$ in degree $0$, we have $\mathrm{ch}(\mathcal{H}) = \Omega \cdot (1 + \mathrm{ch}(\mathcal{H}'))$, where $\Omega = \sum_n h_n(xy)$, the Cauchy kernel, is the part of degree $0$:

```python
Omega = chH_series[0]
Omega
```

Equality of lazy series is not decidable, so we compare them coefficient by coefficient:

```python
LS = Omega.parent()
all(chH_series[d][N] == (Omega * LS(chH_prime_table[d]))[N] for d in range(1, 5) for N in range(d, d + 7))
```
